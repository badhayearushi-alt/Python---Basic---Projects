```python
import pandas as pd
import nltk

# Download stopwords
nltk.download("stopwords")

from nltk.corpus import stopwords

# Load dataset
df = pd.read_csv(
    "test.txt",
    sep=";",
    names=["text", "emotions"],
    header=None
)

# English stopwords
stop_words = set(
    stopwords.words("english")
)

# Function to remove stopwords
def remove_stopwords(text):
    words = text.split()
    cleaned_words = [
        word for word in words
        if word.lower() not in stop_words
    ]
    return " ".join(cleaned_words)

# Apply function
df["stopword_cleaned"] = df["text"].apply(
    remove_stopwords
)

# Display 5 examples
for i in range(5):
    print("\nExample", i + 1)
    print("Before:")
    print(df["text"].iloc[i])
    print("After:")
    print(df["stopword_cleaned"].iloc[i])
```

```python
import pandas as pd
import re
import string
import nltk

# Download stopwords
nltk.download("stopwords")

from nltk.corpus import stopwords

# Load dataset
df = pd.read_csv(
    "test.txt",
    sep=";",
    names=["text", "emotions"],
    header=None
)

# English stopwords
stop_words = set(
    stopwords.words("english")
)

# Complete cleaning function
def clean_text(text):
    # 1. Lowercase
    text = text.lower()

    # 2. Remove punctuation
    text = text.translate(
        str.maketrans("", "", string.punctuation)
    )

    # 3. Remove numbers
    text = re.sub(r"\d+", "", text)

    # 4. Remove special characters
    text = text.encode(
        "ascii", "ignore"
    ).decode("ascii")

    # 5. Remove stopwords
    words = text.split()
    words = [
        word for word in words
        if word not in stop_words
    ]

    return " ".join(words)


# Apply complete pipeline
df["cleaned_text"] = df["text"].apply(
    clean_text
)

# Display 10 samples
print(
    df[["text", "cleaned_text"]]
    .head(10)
    .to_string(index=False)
)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv(
    "test.txt",
    sep=";",
    names=["text", "emotions"],
    header=None
)

# Simple cleaning for length analysis
df["cleaned_text"] = (
    df["text"]
    .str.lower()
)

# Count words
df["text_length"] = df["cleaned_text"].apply(
    lambda x: len(x.split())
)

# Display lengths
print("Text Length:")
print(
    df[["cleaned_text", "text_length"]]
    .head(10)
)

# Statistics
print("\nAverage Text Length:")
print(round(df["text_length"].mean(), 2))

print("\nMinimum Text Length:")
print(df["text_length"].min())

print("\nMaximum Text Length:")
print(df["text_length"].max())

# Histogram
plt.figure(figsize=(10, 5))

plt.hist(
    df["text_length"],
    bins=30,
    edgecolor="black"
)

plt.title("Histogram of Text Lengths")
plt.xlabel("Number of Words")
plt.ylabel("Frequency")

plt.show()

# Observations
print("\nObservations:")
print("1. Text length represents the number of words in each text.")
print("2. Text lengths are different for different sentences.")
print("3. The histogram shows the distribution of text lengths.")
```

```python
import pandas as pd
import re
import string
import nltk
from sklearn.preprocessing import LabelEncoder

# Download stopwords
nltk.download("stopwords")

from nltk.corpus import stopwords


# ============================================================
# 1. LOAD DATA
# ============================================================

df = pd.read_csv(
    "test.txt",
    sep=";",
    names=["text", "emotions"],
    header=None
)

print("Dataset Loaded Successfully!")
print("Shape:", df.shape)


# ============================================================
# 2. ENCODE EMOTION LABELS
# ============================================================

encoder = LabelEncoder()

df["emotion_encoded"] = encoder.fit_transform(
    df["emotions"]
)

print("\nEmotion Mapping:")

mapping = dict(
    zip(
        encoder.classes_,
        encoder.transform(encoder.classes_)
    )
)

print(mapping)


# ============================================================
# 3. COMPLETE TEXT CLEANING
# ============================================================

stop_words = set(
    stopwords.words("english")
)


def clean_text(text):

    # Lowercase
    text = text.lower()

    # Remove punctuation
    text = text.translate(
        str.maketrans("", "", string.punctuation)
    )

    # Remove numbers
    text = re.sub(
        r"\d+",
        "",
        text
    )

    # Keep ASCII characters only
    text = text.encode(
        "ascii",
        "ignore"
    ).decode("ascii")

    # Remove stopwords
    words = text.split()

    words = [
        word
        for word in words
        if word not in stop_words
    ]

    return " ".join(words)


df["cleaned_text"] = df["text"].apply(
    clean_text
)


# ============================================================
# 4. SAVE CLEANED DATASET
# ============================================================

df.to_csv(
    "cleaned_emotions.csv",
    index=False
)

print("\nCleaned dataset saved as:")
print("cleaned_emotions.csv")


# ============================================================
# 5. VALUE COUNTS OF EACH EMOTION
# ============================================================

print("\nEmotion Value Counts:")

print(
    df["emotions"].value_counts()
)


# ============================================================
# DISPLAY FINAL DATA
# ============================================================

print("\nFinal Dataset:")

print(
    df[
        [
            "text",
            "emotions",
            "emotion_encoded",
            "cleaned_text"
        ]
    ].head(10).to_string(index=False)
)
```
