```python
sns.countplot(x=df["channel_status"])
plt.xticks(rotation=45)
plt.show()
```

```python
# 13. Heatmap
plt.figure(figsize=(8,6))
sns.heatmap(df.select_dtypes(include=np.number).corr(), annot=True, cmap="coolwarm")
plt.show()
```
