```python
import pandas as pd
import nltk
import string
import re

# Download stopwords
nltk.download("stopwords")

from nltk.corpus import stopwords

# Load Dataset
df = pd.read_csv(
    "test.txt",
    sep=";",
    names=["text", "emotions"],
    header=None
)

# English stopwords
stop_words = set(stopwords.words("english"))

# Complete cleaning function for Q7
def clean_text(text):
    # Lowercase
    text = text.lower()

    # Remove punctuation
    text = text.translate(
        str.maketrans("", "", string.punctuation)
    )

    # Remove numbers
    text = re.sub(r'\d+', '', text)

    # Keep only ASCII characters
    text = ''.join(
        char for char in text
        if ord(char) < 128
    )

    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    # Remove stopwords
    words = text.split()

    words = [
        word for word in words
        if word not in stop_words
    ]

    return ' '.join(words)


# Apply cleaning
df["cleaned_text"] = df["text"].apply(clean_text)

# Display 5 samples
print("AFTER STOPWORD REMOVAL:")

for text in df["cleaned_text"].head(5):
    print(text)
```

```python
import pandas as pd
import string
import re
import nltk

# Download stopwords
nltk.download("stopwords")

from nltk.corpus import stopwords

# Load Dataset
df = pd.read_csv(
    "test.txt",
    sep=";",
    names=["text", "emotions"],
    header=None
)

# Stopwords
stop_words = set(stopwords.words("english"))

# Complete cleaning function
def clean_text(text):
    # 1. Lowercase
    text = text.lower()

    # 2. Remove punctuation
    text = text.translate(
        str.maketrans("", "", string.punctuation)
    )

    # 3. Remove numbers
    text = re.sub(r'\d+', '', text)

    # 4. Remove emojis and special characters
    text = ''.join(
        char for char in text
        if ord(char) < 128
    )

    # 5. Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    # 6. Remove stopwords
    words = text.split()

    words = [
        word for word in words
        if word not in stop_words
    ]

    return ' '.join(words)


# Apply complete pipeline
df["cleaned_text"] = df["text"].apply(clean_text)

# Display original and cleaned text
print("ORIGINAL TEXT vs CLEANED TEXT\n")

for i in range(10):
    print("Original :", df.loc[i, "text"])
    print("Cleaned  :", df.loc[i, "cleaned_text"])
    print("-" * 80)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load Dataset
df = pd.read_csv(
    "test.txt",
    sep=";",
    names=["text", "emotions"],
    header=None
)

# Import libraries
import string
import re
import nltk

nltk.download("stopwords")

from nltk.corpus import stopwords

stop_words = set(stopwords.words("english"))

# Cleaning function
def clean_text(text):
    # Lowercase
    text = text.lower()

    # Remove punctuation
    text = text.translate(
        str.maketrans("", "", string.punctuation)
    )

    # Remove numbers
    text = re.sub(r'\d+', '', text)

    # Remove special characters
    text = ''.join(
        char for char in text
        if ord(char) < 128
    )

    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    # Remove stopwords
    words = text.split()

    words = [
        word for word in words
        if word not in stop_words
    ]

    return ' '.join(words)


# Apply cleaning
df["cleaned_text"] = df["text"].apply(clean_text)

# Calculate number of words
df["text_length"] = df["cleaned_text"].apply(
    lambda x: len(x.split())
)

# Display lengths
print("Text Lengths:")
print(df[["cleaned_text", "text_length"]].head(10))

# Average
average_length = df["text_length"].mean()

# Minimum
minimum_length = df["text_length"].min()

# Maximum
maximum_length = df["text_length"].max()

print("\nAverage Length:", average_length)
print("Minimum Length:", minimum_length)
print("Maximum Length:", maximum_length)

# Histogram
plt.figure(figsize=(8, 5))

plt.hist(
    df["text_length"],
    bins=20,
    edgecolor="black"
)

plt.title("Histogram of Cleaned Text Lengths")
plt.xlabel("Number of Words")
plt.ylabel("Frequency")
plt.show()

# Observation
print("\nObservation:")
print("The histogram shows the distribution of word counts")
print("after text cleaning.")
print("Cleaning removes unnecessary punctuation, numbers,")
print("special characters and stopwords.")
```

```python
# ==========================================
# Q10 - MINI PROJECT
# FULL TEXT PREPROCESSING PIPELINE
# ==========================================

import pandas as pd
import string
import re
import nltk

from sklearn.preprocessing import LabelEncoder

# ------------------------------------------
# 1. Download NLTK Stopwords
# ------------------------------------------

nltk.download("stopwords")

from nltk.corpus import stopwords

# ------------------------------------------
# 2. Load Dataset
# ------------------------------------------

df = pd.read_csv(
    "test.txt",
    sep=";",
    names=["text", "emotions"],
    header=None
)

print("Dataset Loaded Successfully!")
print("Shape:", df.shape)

# ------------------------------------------
# 3. Encode Emotion Labels
# ------------------------------------------

label_encoder = LabelEncoder()

df["emotion_encoded"] = label_encoder.fit_transform(
    df["emotions"]
)

# Display mapping
emotion_mapping = dict(
    zip(
        label_encoder.classes_,
        label_encoder.transform(label_encoder.classes_)
    )
)

print("\nEmotion Mapping:")
print(emotion_mapping)

# ------------------------------------------
# 4. Stopwords
# ------------------------------------------

stop_words = set(stopwords.words("english"))

# ------------------------------------------
# 5. Complete Cleaning Function
# ------------------------------------------

def clean_text(text):
    # Lowercase
    text = text.lower()

    # Remove punctuation
    text = text.translate(
        str.maketrans("", "", string.punctuation)
    )

    # Remove numbers
    text = re.sub(r'\d+', '', text)

    # Remove emojis and special characters
    text = ''.join(
        char for char in text
        if ord(char) < 128
    )

    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    # Remove stopwords
    words = text.split()

    words = [
        word for word in words
        if word not in stop_words
    ]

    return ' '.join(words)


# ------------------------------------------
# 6. Apply Complete Cleaning
# ------------------------------------------

df["cleaned_text"] = df["text"].apply(clean_text)

print("\nCleaning Completed!")

# ------------------------------------------
# 7. Display Original and Cleaned Text
# ------------------------------------------

print("\nOriginal vs Cleaned Text:\n")

for i in range(10):
    print("Original :", df.loc[i, "text"])
    print("Cleaned  :", df.loc[i, "cleaned_text"])
    print("Emotion  :", df.loc[i, "emotions"])
    print("-" * 80)

# ------------------------------------------
# 8. Save Cleaned Dataset
# ------------------------------------------

df.to_csv(
    "cleaned_emotions.csv",
    index=False
)

print("\nCleaned dataset saved as:")
print("cleaned_emotions.csv")

# ------------------------------------------
# 9. Value Counts of Each Emotion
# ------------------------------------------

print("\nValue Counts of Each Emotion:")

print(df["emotions"].value_counts())

# ------------------------------------------
# 10. Final Dataset Preview
# ------------------------------------------

print("\nFinal Dataset:")
print(df.head(10))
```
