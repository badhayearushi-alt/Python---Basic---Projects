```python
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

best_acc = 0
best_param = None

for C in [1, 10, 20]:
    for kernel in ['linear', 'rbf']:
        svm = SVC(C=C, kernel=kernel, random_state=42)
        svm.fit(x_train, y_train)

        y_pred = svm.predict(x_test)
        acc = accuracy_score(y_test, y_pred)

        print(f"C = {C}  Kernel = {kernel}  Accuracy = {round(acc,2)}")

        if acc > best_acc:
            best_acc = acc
            best_param = (C, kernel)

print("\nBest Parameters =", best_param)
print("Best Accuracy =", round(best_acc,2))
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVC

param_grid = {
    'C': [1, 10, 20],
    'kernel': ['linear', 'rbf']
}

grid = GridSearchCV(
    SVC(),
    param_grid,
    cv=5,
    scoring='accuracy'
)

grid.fit(x_train, y_train)

# Shows the GridSearchCV object
grid
```

```python
import pandas as pd

results = pd.DataFrame(grid.cv_results_)

print(results[['param_C', 'param_kernel', 'mean_test_score']])

print("\nBest Parameters :", grid.best_params_)
print("Best Score :", round(grid.best_score_, 2))
```

```python
from sklearn.model_selection import RandomizedSearchCV
from sklearn.svm import SVC

param_dist = {
    'C': [1, 10, 20],
    'kernel': ['linear', 'rbf']
}

rcv = RandomizedSearchCV(
    SVC(),
    param_distributions=param_dist,
    n_iter=5,
    cv=5,
    random_state=42
)

rcv.fit(x_train, y_train)

# Display RandomizedSearchCV object
rcv
```

```python
import pandas as pd

result = pd.DataFrame(rcv.cv_results_)

print(result[['param_C','param_kernel','mean_test_score']])

print("\nBest Parameters :", rcv.best_params_)
print("Best Score :", round(rcv.best_score_,2))
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV

# Parameter Grid
param_grid = {
    'n_estimators': [10, 50, 100],
    'max_depth': [2, 3, 5]
}

# Random Forest Model
rf = RandomForestClassifier(random_state=42)

# Grid Search
grid_rf = GridSearchCV(
    estimator=rf,
    param_grid=param_grid,
    cv=5,
    scoring='accuracy'
)

# Train
grid_rf.fit(x_train, y_train)

# Output
print("Best Parameters :", grid_rf.best_params_)
print("Best Score :", grid_rf.best_score_)
```

```python
import pandas as pd
import joblib

# Get the best Random Forest model from GridSearchCV
fitted_rf = grid_rf.best_estimator_

# Model comparison
models = {
    "SVM": svm.score(x_test, y_test),
    "Random Forest": fitted_rf.score(x_test, y_test),
    "AdaBoost": ada.score(x_test, y_test),
    "Gradient Boosting": gb.score(x_test, y_test),
    "XGBoost": xgb.score(x_test, y_test)
}

df = pd.DataFrame({
    "Model": models.keys(),
    "Accuracy": models.values()
})

print(df)

# Best model
best_model_name = df.loc[df["Accuracy"].idxmax(), "Model"]
print("\nBest Model")
print(df.loc[df["Accuracy"].idxmax()])

# Save best model
if best_model_name == "SVM":
    joblib.dump(svm, "best_model.pkl")
elif best_model_name == "Random Forest":
    joblib.dump(fitted_rf, "best_model.pkl")
elif best_model_name == "AdaBoost":
    joblib.dump(ada, "best_model.pkl")
elif best_model_name == "Gradient Boosting":
    joblib.dump(gb, "best_model.pkl")
else:
    joblib.dump(xgb, "best_model.pkl")

print("\nModel Saved Successfully")
```

```python
import pandas as pd

df = pd.read_csv("iris.csv")

print("Dataset loaded successfully")
```

```python
from sklearn.model_selection import train_test_split

X = df.iloc[:, :-1]
y = df.iloc[:, -1]

x_train, x_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.33,
    random_state=42
)

print("Dataset trained successfully")
```

```python
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

print("Manual Search")

for C in [1, 10, 20]:
    for kernel in ['linear', 'rbf']:
        svm = SVC(C=C, kernel=kernel)
        svm.fit(x_train, y_train)

        pred = svm.predict(x_test)
        acc = accuracy_score(y_test, pred)

        print(f"C={C}, Kernel={kernel}, Accuracy={acc:.4f}")
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVC

param_grid = {
    'C': [1, 10, 20],
    'kernel': ['linear', 'rbf']
}

grid = GridSearchCV(
    SVC(),
    param_grid,
    cv=5
)

grid.fit(x_train, y_train)

grid
```

```python
 print("Grid Best Parameters:", grid.best_params_)
 print("Grid Best Score:", grid.best_score_)
```

```python
from sklearn.model_selection import RandomizedSearchCV
from sklearn.svm import SVC

random = RandomizedSearchCV(
    SVC(),
    param_distributions=param_grid,
    n_iter=5,
    cv=5,
    random_state=42
)

random.fit(x_train, y_train)

random
```

```python
 print("Random Best Parameters:", random.best_params_)
 print("Random Best Score:", random.best_score_)
```
