```python
import matplotlib.pyplot as plt
import seaborn as sns

cols = ['age','bmi','children','expenses']

for col in cols:
    plt.figure(figsize=(5,4))
    sns.histplot(df[col], kde=True)
    plt.title(f"Distribution of {col}")
    plt.show()
```

```python
cols = ['sex','smoker','region']

for col in cols:
    plt.figure(figsize=(5,4))
    sns.countplot(x=df[col])
    plt.title(f"Count Plot of {col}")
    plt.show()
```

```python
plt.figure(figsize=(6,5))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
print("Average:", df['expenses'].mean())
print("Maximum:", df['expenses'].max())
print("Minimum:", df['expenses'].min())

print(df.groupby("smoker")["expenses"].mean())
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x="smoker", y="expenses", data=df)
plt.show()

plt.figure(figsize=(6,4))
sns.boxplot(x="sex", y="expenses", data=df)
plt.show()
```

```python
# Q10 - Mini Project / Analysis Summary

import pandas as pd

# Load the dataset
df = pd.read_csv("insurance.csv")

# If your dataset uses 'expenses' instead of 'charges'
if "expenses" in df.columns:
    df.rename(columns={"expenses": "charges"}, inplace=True)

    # 1. Average Age and BMI
    avg_age = df["age"].mean()
    avg_bmi = df["bmi"].mean()

    print("1. Average Age:", round(avg_age, 2))
    print("Average BMI:", round(avg_bmi, 2))

    # 2. Smoking effect on insurance charges
    print("\n2. Average Charges by Smoker:")
    print(df.groupby("smoker")["charges"].mean())

    # 3. Region with the highest number of customers
    region_count = df["region"].value_counts()

    print("\n3. Number of Customers by Region:")
    print(region_count)

    print("\nRegion with Highest Customers:", region_count.idxmax())

    # 4. Observations
    print("\n4. Observations:")
    print("- Smokers have much higher insurance charges than non-smokers.")
    print("-", region_count.idxmax(), "has the highest number of customers.")
    print("- Average Age =", round(avg_age, 2))
    print("- Average BMI =", round(avg_bmi, 2))
    print("- Insurance charges contain some outliers.")
```
